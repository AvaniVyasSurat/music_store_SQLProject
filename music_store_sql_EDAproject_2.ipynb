## Write query to return the email, first name, last name, & Genre of all Rock Music listeners. Return your list ordered alphabetically by email starting with A.

```python
#/*Method 1 */

SELECT DISTINCT email,first_name, last_name
FROM customer
JOIN invoice ON customer.customer_id = invoice.customer_id
JOIN invoiceline ON invoice.invoice_id = invoiceline.invoice_id
WHERE track_id IN(
	SELECT track_id FROM track
	JOIN genre ON track.genre_id = genre.genre_id
	WHERE genre.name LIKE 'Rock'
)
ORDER BY email;

#/* Method 2 */
SELECT DISTINCT email AS Email,first_name AS FirstName, last_name AS LastName, genre.name AS Name
FROM customer
JOIN invoice ON invoice.customer_id = customer.customer_id
JOIN invoiceline ON invoiceline.invoice_id = invoice.invoice_id
JOIN track ON track.track_id = invoiceline.track_id
JOIN genre ON genre.genre_id = track.genre_id
WHERE genre.name LIKE 'Rock'
ORDER BY email;
```

### Explanation: Rock Music Listeners Query

**What this query does:**
This query retrieves a unique list of emails, first names, last names, and the genre name ("Rock") of all customers who have purchased at least one Rock music track. The final list is sorted alphabetically by the customer's email address.

**How it works (Method 2):**
1. **`FROM customer`**: Starts with the `customer` table containing customer profile details.
2. **`JOIN` chain**:
   * Connects `customer` to `invoice` using `customer_id` to find purchase records.
   * Connects `invoice` to `invoiceline` using `invoice_id` to get individual items purchased.
   * Connects `invoiceline` to `track` using `track_id` to identify which songs were bought.
   * Connects `track` to `genre` using `genre_id` to find the genre of each song.
3. **`WHERE genre.name LIKE 'Rock'`**: Filters the records to keep only the purchases of tracks belonging to the 'Rock' genre.
4. **`SELECT DISTINCT`**: Ensures that even if a customer bought multiple Rock songs, their name and email appear only once in the final output.
5. **`ORDER BY email`**: Sorts the final output alphabetically from A to Z by email.

## Let's invite the artists who have written the most rock music in our dataset. Write a query that returns the Artist name and total track count of the top 10 rock bands.

```python
SELECT artist.artist_id, artist.name,COUNT(artist.artist_id) AS number_of_songs
FROM track
JOIN album ON album.album_id = track.album_id
JOIN artist ON artist.artist_id = album.artist_id
JOIN genre ON genre.genre_id = track.genre_id
WHERE genre.name LIKE 'Rock'
GROUP BY artist.artist_id
ORDER BY number_of_songs DESC
LIMIT 10;
```

### Explanation: Top 10 Rock Bands Query

**What this query does:**
This query identifies the top 10 artists/bands in the database who have written the most tracks classified under the "Rock" genre. It displays their unique Artist ID, Name, and the total count of their Rock songs, sorted from highest to lowest.

**How it works:**
1. **`FROM track`**: Starts with the `track` table containing all songs.
2. **`JOIN` chain**:
   * Connects `track` to `album` via `album_id` to see which album each song belongs to.
   * Connects `album` to `artist` via `artist_id` to find who created that album.
   * Connects `track` to `genre` via `genre_id` to filter by music style.
3. **`WHERE genre.name LIKE 'Rock'`**: Filters the dataset to include only Rock tracks.
4. **`GROUP BY artist.artist_id`**: Groups the filtered tracks by artist so we can count songs per artist.
5. **`COUNT(artist.artist_id) AS number_of_songs`**: Counts the number of occurrences of each artist in the filtered list (which represents their total number of Rock tracks).
6. **`ORDER BY number_of_songs DESC`**: Sorts the artists in descending order so those with the most songs appear first.
7. **`LIMIT 10`**: Restricts the final output to only show the top 10 records.

## Return all the track names that have a song length longer than the average song length. Return the Name and Milliseconds for each track. Order by the song length with the longest songs listed first.

```python

SELECT name,miliseconds
FROM track
WHERE miliseconds > (
	SELECT AVG(miliseconds) AS avg_track_length
	FROM track )
ORDER BY miliseconds DESC;
```

### Explanation: Above Average Song Length Query

**What this query does:**
This query returns a list of all song names and their lengths (in milliseconds) that are longer than the average duration of all tracks in the database. The results are ordered with the longest tracks at the very top.

**How it works:**
1. **Subquery (`SELECT AVG(miliseconds) FROM track`)**: This inner query calculates the average length of all songs in the entire database.
2. **`WHERE miliseconds > (...)`**: The outer query compares the duration of each individual track against the average calculated by the subquery, keeping only the ones that exceed it.
3. **`SELECT name, miliseconds`**: Retrieves the name and duration of those qualifying tracks.
4. **`ORDER BY miliseconds DESC`**: Sorts the list from longest to shortest duration.
