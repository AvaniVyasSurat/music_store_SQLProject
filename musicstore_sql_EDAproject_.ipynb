## Q1: Who is the senior most employee based on job title?

##
SELECT title, last_name, first_name
FROM employee
ORDER BY levels DESC
LIMIT 1

### **Explanation & Answer for Q1**
* **Answer:** This query identifies the senior-most employee by looking at the highest job level.
* **How it works:**
  1. It selects the `title`, `last_name`, and `first_name` from the `employee` table.
  2. It orders the records by the `levels` column in descending order (`DESC`), so that the employee with the highest organizational level is at the very top.
  3. It uses `LIMIT 1` to return only that single top record.

## Q2: Which countries have the most Invoices?

##
SELECT COUNT(*) AS c, billing_country
FROM invoice
GROUP BY billing_country
ORDER BY c DESC

### **Explanation & Answer for Q2**
* **Answer:** This query lists countries ranked by the total number of invoices they have generated, showing which countries have the most transactions.
* **How it works:**
  1. It groups all records in the `invoice` table by their `billing_country` using `GROUP BY`.
  2. For each country group, `COUNT(*)` counts the total number of invoices and labels the column as `c`.
  3. Finally, it sorts the list in descending order (`ORDER BY c DESC`) so that the country with the most invoices appears first.

## Q3: What are top 3 values of total invoice?

##
SELECT total
FROM invoice
ORDER BY total DESC

### **Explanation & Answer for Q3**
* **Answer:** This query finds the highest invoice amounts recorded in the dataset.
* **How it works:**
  1. It selects the `total` column from the `invoice` table.
  2. It sorts the totals in descending order (`ORDER BY total DESC`) to place the largest amounts at the top.
  * *Note: To strictly get the "top 3" values, adding `LIMIT 3` to this query would ensure only the top 3 rows are returned.*

## Q4: Which city has the best customers? We would like to throw a promotional Music Festival in the city we made the most money. Write a query that returns one city that has the highest sum of invoice totals. Return both the city name & sum of all invoice totals



##
SELECT billing_city,SUM(total) AS InvoiceTotal
FROM invoice
GROUP BY billing_city
ORDER BY InvoiceTotal DESC
LIMIT 1;

### **Explanation & Answer for Q4**
* **Answer:** This query identifies the single city that generated the highest total revenue, which would be the best location to host the Music Festival.
* **How it works:**
  1. It groups invoice records by `billing_city`.
  2. It calculates the sum of all invoice totals for each city using `SUM(total)` and names it `InvoiceTotal`.
  3. It sorts these sums from highest to lowest (`ORDER BY InvoiceTotal DESC`).
  4. `LIMIT 1` filters the final output to display only the single city with the highest cumulative revenue.

## Q5: Who is the best customer? The customer who has spent the most money will be declared the best customer. Write a query that returns the person who has spent the most money.

##
SELECT customer.customer_id, first_name, last_name, SUM(total) AS total_spending
FROM customer
JOIN invoice ON customer.customer_id = invoice.customer_id
GROUP BY customer.customer_id
ORDER BY total_spending DESC
LIMIT 1;

### **Explanation & Answer for Q5**
* **Answer:** This query finds the best customer by identifying who has spent the most money across all their invoices.
* **How it works:**
  1. It joins the `customer` table with the `invoice` table using the common `customer_id` field to link customer details with their purchases.
  2. It groups the results by `customer.customer_id`.
  3. It sums up all invoice totals for each customer (`SUM(total)`) and labels it `total_spending`.
  4. It orders the customers by their total spending in descending order (`ORDER BY total_spending DESC`) and uses `LIMIT 1` to return only the customer who spent the most.
